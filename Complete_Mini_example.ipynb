```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
 "Age": [21, 22, 25, 28, 30, 32, 35, 40],
 "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 9],
 "Marks": [45, 50, 55, 62, 68, 74, 82, 90]
}
df = pd.DataFrame(data)
```

```python
print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
```

```python
average_marks = np.mean(df["Marks"])
median_marks = np.median(df["Marks"])
std_marks = np.std(df["Marks"])
print("Average:", average_marks)
print("Median:", median_marks)
print("Standard Deviation:", std_marks)
```

```python
plt.scatter(df["Study_Hours"], df["Marks"])
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
plt.hist(df["Marks"])
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.boxplot(df["Marks"])
plt.title("Marks Boxplot")
plt.ylabel("Marks")
plt.show()
```
