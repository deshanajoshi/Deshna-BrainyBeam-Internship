```python
import pandas as pd
import numpy as np
```

TASK 1: LOAD THE DATASET

```python
df = pd.read_csv("student_data.csv")

print("Dataset Loaded Successfully")
print(df)
```

TASK 2: UNDERSTAND THE DATASET

```python
print("\nFirst 5 Rows:")
print(df.head())
```

```python
# Number of rows and columns
print("\nShape:")
print(df.shape)


print("\nColumn Names:")
print(df.columns)


print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()


print("\nStatistical Summary:")
print(df.describe())
```

 TASK 3: ANALYZE CATEGORICAL DATA

```python
print("\nUnique Gender Values:")
print(df["Gender"].unique())

print("\nNumber of Unique Genders:")
print(df["Gender"].nunique())

print("\nGender Counts:")
print(df["Gender"].value_counts())
```

TASK 4: CHECK DATA QUALITY

```python
print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nDuplicate Student IDs:")
print(df["Student_ID"].duplicated().sum())

print("\nInvalid Age:")
print((~df["Age"].between(1, 100)).sum())

print("\nInvalid Study Hours:")
print((~df["Study_Hours"].between(0, 24)).sum())

print("\nInvalid Attendance:")
print((~df["Attendance"].between(0, 100)).sum())

print("\nInvalid Marks:")
print((~df["Marks"].between(0, 100)).sum())

print("\nInvalid Assignments:")
print((df["Assignments"] < 0).sum())

print("\nInvalid Gender:")
print((~df["Gender"].isin(["Male", "Female"])).sum())
```

 TASK 5: FILTER STUDENTS

```python
print("\nStudents with Marks > 80:")
print(df[df["Marks"] > 80])

print("\nStudents with Attendance > 75:")
print(df[df["Attendance"] > 75])

print("\nStudents with Study Hours > 5:")
print(df[df["Study_Hours"] > 5])
```

TASK 6: CALCULATE STATISTICS

```python
print("\nAverage Marks:")
print(df["Marks"].mean())

print("\nMedian Marks:")
print(df["Marks"].median())

print("\nAverage Study Hours:")
print(df["Study_Hours"].mean())

print("\nHighest Marks:")
print(df["Marks"].max())

print("\nLowest Marks:")
print(df["Marks"].min())
```

TASK 7: GROUP THE DATA

```python
print("\nAverage Marks by Gender:")
print(df.groupby("Gender")["Marks"].mean())

# Create attendance groups
df["Attendance_Group"] = pd.cut(
    df["Attendance"],
    bins=[-1, 60, 75, 90, 100],
    labels=["Low", "Medium", "High", "Excellent"]
)

print("\nAverage Marks by Attendance Group:")
print(df.groupby(
    "Attendance_Group",
    observed=False
)["Marks"].mean())
```

TASK 8: CREATE A NEW FEATURE

```python
# Passed if marks are 40 or above
df["Passed"] = df["Marks"] >= 40

print("\nDataset with Passed Column:")
print(df.head())

print("\nPass/Fail Counts:")
print(df["Passed"].value_counts())

df.to_csv("student_data_updated.csv", index=False)

print("\nUpdated dataset saved successfully!")
```
