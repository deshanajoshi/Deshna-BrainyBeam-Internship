```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/student_data.csv")
```

```python
plt.hist(df["Marks"], bins=10, edgecolor="black")

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.scatter(df["Study_Hours"], df["Marks"])

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
plt.scatter(df["Attendance"], df["Marks"])

plt.title("Attendance vs Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")

plt.show()
```

```python
plt.boxplot(df["Marks"])

plt.title("Boxplot of Marks")
plt.ylabel("Marks")

plt.show()
```

```python
avg_marks = df.groupby("Gender")["Marks"].mean()

plt.bar(avg_marks.index, avg_marks.values)

plt.title("Average Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Marks")

plt.show()
```

```python
avg_marks = df.groupby("Study_Hours")["Marks"].mean()

plt.plot(avg_marks.index, avg_marks.values, marker="o")

plt.title("Average Marks by Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Average Marks")

plt.show()
```
