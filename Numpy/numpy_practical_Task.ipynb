```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/student_data.csv")

print(df.head())
```

1. Convert a column to a NumPy array

```python
marks = df["Marks"].to_numpy()
print(marks)
```

2. Calculate mean

```python
print(np.mean(marks))
```

3. Calculate median

```python
print(np.median(marks))
```

4. Calculate standard deviation

```python
print(np.std(marks))
```

5. Find minimum and maximum

```python
print(np.min(marks))
print(np.max(marks))
```

6. Apply a mathematical operation

```python
print(marks + 5)
```

7. Filter values using a condition

```python
print(marks[marks > 75])
```

8. Reshape a small NumPy array

```python
arr = np.array([1, 2, 3, 4, 5, 6])
print(arr.reshape(2, 3))
```

9. Experiment with arange()

```python
print(np.arange(1, 21, 2))
```

10. Experiment with linspace()

```python
print(np.linspace(0, 100, 6))
```
